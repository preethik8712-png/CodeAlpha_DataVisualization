```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

df.head()
```

```python
survival_count = df["Survived"].value_counts()

plt.figure(figsize=(7,5))
plt.bar(["Not Survived", "Survived"], survival_count.values)

plt.title("Titanic Survival Count")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
gender_survival = pd.crosstab(df["Sex"], df["Survived"])

gender_survival.plot(kind="bar", figsize=(7,5))

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.legend(["Not Survived", "Survived"])

plt.show()
```

```python
class_survival = pd.crosstab(df["Pclass"], df["Survived"])

class_survival.plot(kind="bar", figsize=(7,5))

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.legend(["Not Survived", "Survived"])

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Age"].dropna(), bins=20)

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Fare"], bins=20)

plt.title("Fare Distribution of Passengers")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Age"], df["Fare"], alpha=0.5)

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```
